# **Planteamiento de hipótesis**

```python
# Separar las ventas reales de cada marca

ventas_bosi = df[df["MARCA"] == "BOSI"]["REAL"]
ventas_bambino = df[df["MARCA"] == "BAMBINO"]["REAL"]

print("Ventas BOSI:")
print(ventas_bosi)

print("\nVentas BAMBINO:")
print(ventas_bambino)
```

```python
# Número de observaciones por marca

print("Número de registros BOSI:", len(ventas_bosi))
print("Número de registros BAMBINO:", len(ventas_bambino))
```

# **Ejecutar la prueba t**

```python
# Prueba t de Student para muestras independientes

from scipy.stats import ttest_ind

resultado_t = ttest_ind(
    ventas_bosi,
    ventas_bambino,
    equal_var=False
)

print("Estadístico t:", resultado_t.statistic)
print("p-valor:", resultado_t.pvalue)
```

# **¿Por qué equal_var=False?**

Estamos utilizando la versión de la prueba t que no supone que las dos poblaciones tengan exactamente la misma varianza. Es una elección más conservadora para comparar dos grupos independientes.

# **Comprobar normalidad**

```python
# Prueba de normalidad de Shapiro-Wilk

from scipy.stats import shapiro

normalidad_bosi = shapiro(ventas_bosi)
normalidad_bambino = shapiro(ventas_bambino)

print("Prueba de normalidad - BOSI")
print("Estadístico:", normalidad_bosi.statistic)
print("p-valor:", normalidad_bosi.pvalue)

print("\nPrueba de normalidad - BAMBINO")
print("Estadístico:", normalidad_bambino.statistic)
print("p-valor:", normalidad_bambino.pvalue)
```

# **Comparar los promedios**

```python
# Estadísticas principales de las ventas reales por marca

print("BOSI")
print("Promedio:", ventas_bosi.mean())
print("Mediana:", ventas_bosi.median())
print("Desviación estándar:", ventas_bosi.std())

print("\nBAMBINO")
print("Promedio:", ventas_bambino.mean())
print("Mediana:", ventas_bambino.median())
print("Desviación estándar:", ventas_bambino.std())
```

# **Diferencia porcentual entre los promedios**

```python
# Diferencia porcentual entre los promedios de ventas

promedio_bosi = ventas_bosi.mean()
promedio_bambino = ventas_bambino.mean()

diferencia_porcentual = (
    (promedio_bosi - promedio_bambino) / promedio_bambino
) * 100

print("Diferencia porcentual entre los promedios:")
print(f"{diferencia_porcentual:.2f}%")
```

# **Obtener los valores definitivos**

```python
# Resumen estadístico de las ventas reales por marca

resumen = df.groupby("MARCA")["REAL"].agg(
    Media="mean",
    Mediana="median",
    Desviacion_Estandar="std",
    Minimo="min",
    Maximo="max",
    Observaciones="count"
)

display(resumen)
```

```python
# Diferencia absoluta entre los promedios

promedio_bosi = ventas_bosi.mean()
promedio_bambino = ventas_bambino.mean()

diferencia = promedio_bosi - promedio_bambino

print(f"Promedio BOSI: ${promedio_bosi:,.2f}")
print(f"Promedio BAMBINO: ${promedio_bambino:,.2f}")
print(f"Diferencia entre promedios: ${diferencia:,.2f}")
```

# **Tamaño del efecto**

```python
# Cálculo del tamaño del efecto: Cohen's d

n_bosi = len(ventas_bosi)
n_bambino = len(ventas_bambino)

sd_bosi = ventas_bosi.std()
sd_bambino = ventas_bambino.std()

# Desviación estándar combinada
sd_combinada = np.sqrt(
    ((n_bosi - 1) * sd_bosi**2 + (n_bambino - 1) * sd_bambino**2)
    / (n_bosi + n_bambino - 2)
)

# Cohen's d
cohen_d = (promedio_bosi - promedio_bambino) / sd_combinada

print("Tamaño del efecto (Cohen's d):", cohen_d)
```

Se construye una variable de cumplimiento para determinar qué porcentaje de la meta representa la venta real de cada registro.

Un valor de:

- **100%** representa el cumplimiento exacto de la meta.
- **Menor al 100%** representa un resultado inferior a la meta.
- **Mayor al 100%** representa un resultado superior a la meta.

## 7. Análisis estadístico inferencial

Se aplican pruebas estadísticas para determinar si las diferencias y relaciones observadas en los datos cuentan con evidencia estadística suficiente.

Se utiliza un nivel de significancia de:

# **Prueba t pareada: BOSI vs. BAMBINO**

```python
# Organizar los datos para comparar BOSI y BAMBINO en las mismas semanas

datos_pareados = df.pivot(
    index="SEMANA",
    columns="MARCA",
    values="REAL"
).dropna()

datos_pareados
```

```python
# Diferencia de ventas entre BOSI y BAMBINO por semana

diferencias = datos_pareados["BOSI"] - datos_pareados["BAMBINO"]

print("Diferencias de ventas por semana:")
print(diferencias)
```

```python
from scipy.stats import shapiro

normalidad_diferencias = shapiro(diferencias)

print("Prueba de normalidad de las diferencias")
print("Estadístico W:", normalidad_diferencias.statistic)
print("p-valor:", normalidad_diferencias.pvalue)
```

```python
from scipy.stats import ttest_rel

resultado_t_pareada = ttest_rel(
    datos_pareados["BOSI"],
    datos_pareados["BAMBINO"]
)

print("Prueba t de Student para muestras pareadas")
print("Estadístico t:", resultado_t_pareada.statistic)
print("p-valor:", resultado_t_pareada.pvalue)
```

# **Tamaño del efecto**

```python
# Tamaño del efecto para muestras pareadas

cohen_dz = diferencias.mean() / diferencias.std(ddof=1)

print("Tamaño del efecto (Cohen's dz):", cohen_dz)
```

# **Prueba sobre el cumplimiento de la meta**

```python
from scipy.stats import ttest_1samp

for marca in df["MARCA"].unique():

    datos = df.loc[df["MARCA"] == marca, "CUMPLIMIENTO"]

    resultado = ttest_1samp(datos, popmean=100)

    print("\nMarca:", marca)
    print("Promedio de cumplimiento:", datos.mean())
    print("Estadístico t:", resultado.statistic)
    print("p-valor:", resultado.pvalue)
```

# **Regresión lineal: META → REAL**

## 8. Regresión lineal

Se desarrolla un modelo de regresión lineal simple utilizando META como variable explicativa y REAL como variable dependiente.

El objetivo es evaluar en qué medida la relación lineal entre ambas variables permite explicar y estimar las ventas observadas.

```python
from scipy.stats import linregress

modelo = linregress(df["META"], df["REAL"])

print("Pendiente:", modelo.slope)
print("Intercepto:", modelo.intercept)
print("Coeficiente de correlación:", modelo.rvalue)
print("R²:", modelo.rvalue**2)
print("p-valor:", modelo.pvalue)
```

# **Gráfico de regresión**

```python
plt.figure(figsize=(10, 6))

sns.regplot(
    data=df,
    x="META",
    y="REAL",
    scatter_kws={"s": 80}
)

plt.title("Regresión lineal entre META y REAL")
plt.xlabel("Meta de ventas")
plt.ylabel("Ventas reales")

plt.show()
```

# **Ecuación del modelo**

```python
print(
    f"Ecuación del modelo: REAL = "
    f"{modelo.intercept:.2f} + "
    f"({modelo.slope:.4f} × META)"
)
```

## **Predicciones**

```python
df["REAL_PREDICHO"] = modelo.intercept + modelo.slope * df["META"]

df[["SEMANA", "MARCA", "META", "REAL", "REAL_PREDICHO"]]
```

# **Evaluar el modelo**

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

y_real = df["REAL"]
y_predicho = df["REAL_PREDICHO"]

mae = mean_absolute_error(y_real, y_predicho)
mse = mean_squared_error(y_real, y_predicho)
rmse = np.sqrt(mse)
r2 = r2_score(y_real, y_predicho)

print("Evaluación del modelo de regresión lineal")
print("-----------------------------------------")
print(f"MAE:  ${mae:,.2f}")
print(f"MSE:  {mse:,.2f}")
print(f"RMSE: ${rmse:,.2f}")
print(f"R²:   {r2:.4f}")
```
